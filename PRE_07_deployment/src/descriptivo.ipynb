# Análisis descriptivo — House Data

Exploración de las variables usadas por el modelo de precios (`train_model.py`).

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Funciona tanto si el kernel arranca en src/ como en la raíz del repo
CANDIDATES = [Path("../data/house_data.csv"), Path("PRE_07_deployment/data/house_data.csv")]
DATA_PATH = next(path for path in CANDIDATES if path.exists())

FEATURES = ["bedrooms", "bathrooms", "sqft_living", "sqft_lot", "floors", "waterfront", "condition"]
TARGET = "price"

df = pd.read_csv(DATA_PATH)
df.shape
```

```python
df[FEATURES + [TARGET]].head()
```

## Tipos y valores faltantes

```python
df[FEATURES + [TARGET]].isna().sum().to_frame('missing').join(df.dtypes.rename('dtype'))
```

## Estadísticas descriptivas

```python
df[FEATURES + [TARGET]].describe().T
```

## Distribución del precio

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df[TARGET].plot.hist(bins=50, ax=axes[0], title="price")
df[TARGET].apply("log").plot.hist(bins=50, ax=axes[1], title="log(price)")
plt.tight_layout()
plt.show()
```

## Correlación de las variables con el precio

```python
df[FEATURES + [TARGET]].corr()[TARGET].drop(TARGET).sort_values(ascending=False)
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["sqft_living"], df[TARGET], s=3, alpha=0.3)
ax.set_xlabel("sqft_living")
ax.set_ylabel("price")
plt.show()
```

## Precio medio por categoría

```python
for column in ["waterfront", "condition", "floors"]:
    display(df.groupby(column)[TARGET].agg(["count", "mean"]).round(0))
```
